# FEM → Levenberg–Marquardt Inverse Bioheat (Kandlikar-style)

**Method:** Gonzalez-Hernandez, Recinella, Kandlikar *et al.*, *Infrared Physics & Technology* **105** (2020) 103202 — inverse heat-transfer tumour localisation from surface IR.

This is a **fresh module**. It does **not** use Nelder–Mead and does **not** use a bilateral-symmetry cost. Instead:

- **Forward:** steady Pennes bioheat, solved with **FEniCSx** (reusing the proven solver pattern from `run_cohort.run_fea_forward`).
- **Inverse:** **Levenberg–Marquardt** over β = [x_t, y_t, z_t, d], cost = *direct* MSE between computed and measured surface temperature.
- **Detection:** a recovered tumour that lands **outside the breast** (or shrinks to ~0 diameter) ⇒ no tumour (benign).

**This notebook:** Engineering Data → Mesh → BCs → UOQ prior → forward solve → **Levenberg–Marquardt inverse + identifiability** on one patient, all in ANSYS-style documentation order.

> Run in the **`bioheat`** conda env (FEniCSx, dolfinx, gmsh).

## 1 · Engineering Data — material & physiological properties

Values from the Kandlikar paper (Table 3) — note the **physiological blood density and specific heat**. These set the perfusion coefficient $P=\omega\rho_b c_b$ and therefore the thermal penetration depth $\ell=\sqrt{k/P}$. The previous pipeline used `RHO_BLOOD = 1.0`, giving $\ell\approx0.5$ m ($\gg$ breast) so depth was effectively unobservable; the paper's constants give $\ell\approx25$ mm (breast-scale), making depth identifiable.

```python
# Engineering Data (paper, Table 3)
PAPER = dict(
    k_h     = 0.42,      # W/m.K   healthy-tissue conductivity (LM parameter; nominal)
    rho_b   = 1060.0,    # kg/m3   blood density
    c_b     = 3840.0,    # J/kg.K  blood specific heat
    omega_h = 1.8e-4,    # 1/s     healthy perfusion rate
    omega_t = 9.0e-3,    # 1/s     tumour perfusion rate (~50x healthy)
    Q_h     = 450.0,     # W/m3    healthy metabolic heat
    T_a     = 37.0,      # C       arterial / core (chest-wall Dirichlet)
    T_c     = 37.0,      # C       chest-wall temperature
    h       = 13.5,      # W/m2.K  skin convection coefficient
    T_inf   = 21.0,      # C       ambient air
)
PERF_H = PAPER['omega_h'] * PAPER['rho_b'] * PAPER['c_b']
PERF_T = PAPER['omega_t'] * PAPER['rho_b'] * PAPER['c_b']
ell    = (PAPER['k_h'] / PERF_H) ** 0.5

print('Engineering Data')
print(f"  k_h         = {PAPER['k_h']:.2f} W/m.K")
print(f"  perfusion   P_h = w_h*rho_b*c_b = {PERF_H:8.1f} W/m3.K")
print(f"              P_t = w_t*rho_b*c_b = {PERF_T:8.1f} W/m3.K")
print(f"  penetration depth ell = sqrt(k/P_h) = {ell*1e3:6.1f} mm  (breast-scale -> depth identifiable)")
old_P = 0.0005 * 1.0 * 3600.0
print(f"  [old pipeline P={old_P:.2f} -> ell={(PAPER['k_h']/old_P)**0.5*1e3:6.0f} mm  (>> breast -> depth invisible)]")

def apply_constants(k_h=None):
    # push paper constants into run_cohort so its proven solver/masks use them
    R.K_TISSUE   = float(k_h) if k_h is not None else PAPER['k_h']
    R.PERFUSION  = PERF_H
    R.T_ARTERIAL = PAPER['T_a']
    R.Q_METAB    = PAPER['Q_h']
    R.H_CONV     = PAPER['h']
    R.T_AIR      = PAPER['T_inf']

apply_constants()
print('Constants applied to run_cohort.')
```

```python
# ---- shared 3-D viz helpers (point 2: make the internal tumour visible) ----
# Anterior camera: look along +Z (the anterior skin faces -Z), superior (-Y) points up.
ANT_CAM = dict(eye=dict(x=0.0, y=0.0, z=-2.2), up=dict(x=0.0, y=-1.0, z=0.0))

def sphere_mesh(centre, r, color, n=22, name='tumour', opacity=1.0):
    """Solid sphere as a Mesh3d (convex hull) so a sub-surface tumour stays
    visible through a translucent breast surface."""
    u = np.linspace(0, 2*np.pi, n); v = np.linspace(0, np.pi, n)
    uu, vv = np.meshgrid(u, v)
    x = centre[0] + r*np.cos(uu)*np.sin(vv)
    y = centre[1] + r*np.sin(uu)*np.sin(vv)
    z = centre[2] + r*np.cos(vv)
    return go.Mesh3d(x=x.ravel(), y=y.ravel(), z=z.ravel(),
                     alphahull=0, color=color, opacity=opacity,
                     name=name, showscale=False, flatshading=True)

print('viz helpers ready (ANT_CAM, sphere_mesh)')
```

## 4 · Boundary Conditions (ANSYS-style)

| # | ANSYS BC type | Region | Math form | Value |
|---|---|---|---|---|
| 1 | **Convection** | skin surface | $-k\,\partial T/\partial n = h(T-T_\infty)$ | $h=13.5$ W/m²K, $T_\infty=21$ °C |
| 2 | **Temperature** | chest wall (posterior cap) | $T = T_c$ (Dirichlet) | $37$ °C |
| 3 | **Internal Heat Generation** | healthy tissue | $Q_h$ + perfusion sink | $Q_h=450$ W/m³ |
| 4 | **Internal Heat Generation** | tumour region | Gautherie $Q_t(d)$ + $\omega_t$ | spatial |

The chest wall is the **posterior Z-slab** — the back `CHEST_WALL_FRAC` (10 %) of the mesh depth from the highest Z. This is the cut face where the breast attaches to the chest. A normal-based criterion (`nz > 0`) over-selects because curved upper/lateral skin also has outward normals with positive Z components — it is **not** used here.

```python
# Chest-wall identification: Z-slab (back CHEST_WALL_FRAC of depth from posterior).
_z  = geo['surface_pts'][:, 2]
_z0, _z1 = float(_z.min()), float(_z.max())
CW_FRAC = R.CHEST_WALL_FRAC  = 0.30
cw  = _z > (_z0 + (1.0 - CW_FRAC) * (_z1 - _z0))
faces = geo['faces']

fig = go.Figure(go.Mesh3d(
    x=sp[:,0], y=sp[:,1], z=sp[:,2],
    i=faces[:,0], j=faces[:,1], k=faces[:,2],
    vertexcolor=np.where(cw, 'crimson', 'royalblue'),
    flatshading=False, opacity=1.0,
    lighting=dict(ambient=0.5, diffuse=0.8),
    showscale=False, name=''))

fig.add_trace(go.Scatter3d(
    x=[None], y=[None], z=[None], mode='markers',
    marker=dict(size=10, color='crimson', symbol='square'),
    name=f'Chest wall — Dirichlet T=37 °C  ({int(cw.sum())} verts, back {CW_FRAC*100:.0f}%)'))
fig.add_trace(go.Scatter3d(
    x=[None], y=[None], z=[None], mode='markers',
    marker=dict(size=10, color='royalblue', symbol='square'),
    name=f'Skin — convection  h=13.5 W/m²K, T∞=21 °C  ({int((~cw).sum())} verts)'))

fig.update_layout(
    title=(f'Setup - Boundary Conditions<br>'
           f'<sup>Z-slab threshold: Z > {_z0 + (1.0-CW_FRAC)*(_z1-_z0):.1f} mm  '
           f'(back {CW_FRAC*100:.0f}% ≈ {CW_FRAC*(_z1-_z0):.1f} mm slab)</sup>'),
    scene=dict(aspectmode='data', camera=ANT_CAM),
    legend=dict(x=0.01, y=0.97, bgcolor='rgba(255,255,255,0.8)',
                bordercolor='gray', borderwidth=1),
    height=620, width=880, margin=dict(l=0, r=0, t=60, b=0))
fig.write_html(str(R.RESULTS_DIR / f'{pid}_BC_setup.html'))
fig.show()
print(f'chest-wall verts: {int(cw.sum())} / {len(sp)}  '
      f'(Z > {_z0 + (1.0-CW_FRAC)*(_z1-_z0):.1f} mm)')
```

## 5 · Tumour-location prior — quadrant distribution & synthetic placement

Breast cancer concentrates heavily in the **upper-outer quadrant**, but NOT exclusively — using it as the only synthetic location over-represents one region and biases the inverse test.

**Literature quadrant probabilities** (Senie et al. *Am J Epidemiol* 1983; Kricker et al. *Breast Cancer Res Treat* 2007; Chung et al. *Anticancer Res* 2005):

| Quadrant | Abbrev | Probability |
|---|---|---|
| Upper-outer | UO | 50 % |
| Central / subareolar | C | 18 % |
| Upper-inner | UI | 15 % |
| Lower-outer | LO | 11 % |
| Lower-inner | LI | 6 % |

**Depth from skin surface** (Gautherie & Gros *Radiology* 1980; Ng & Sudharsan *Proc IMechE H* 2001; Kandlikar 2020): thermographically detectable tumours lie **5–25 mm** below the skin. Deeper tumours produce a surface signal below the noise floor of IR (~0.1 °C).

**Axis orientation** — Y points inferiorly (grid_sample y=−1 → top row of thermogram = superior). So *upper = smaller Y*, *outer = away from the lateral midline* (flips per breast). Depth **Z** runs anterior (low) → posterior (high); chest wall is at high Z.

```python
# ── Literature quadrant distribution ──────────────────────────────────────────
# Senie et al. Am J Epidemiol 1983; Kricker et al. Breast Cancer Res Treat 2007
QUAD_DIST  = {'UO': 0.50, 'C': 0.18, 'UI': 0.15, 'LO': 0.11, 'LI': 0.06}
QUAD_NAMES = {'UO': 'Upper-Outer', 'UI': 'Upper-Inner',
              'LO': 'Lower-Outer', 'LI': 'Lower-Inner', 'C': 'Central/subareolar'}

# ── Quadrant classifier (per-breast, Z-slab skin mask) ────────────────────────
def quadrants(geo):
    sp  = geo['surface_pts']
    _z  = sp[:, 2]; _z0, _z1 = float(_z.min()), float(_z.max())
    skin = _z < (_z0 + (1.0 - R.CHEST_WALL_FRAC) * (_z1 - _z0))

    Tm    = geo['T_measured']
    x_mid = float(sp[:, 0].mean())
    hot   = int(np.argmax(np.where(skin, Tm, -1e9)))
    side_pos = bool(sp[hot, 0] > x_mid)

    side   = (sp[:, 0] > x_mid) if side_pos else (sp[:, 0] < x_mid)
    breast = skin & side
    c      = sp[breast].mean(0)   # affected-breast centroid — used for UO/UI/LO/LI axes

    outer = (sp[:, 0] > c[0]) if side_pos else (sp[:, 0] < c[0])
    upper = sp[:, 1] < c[1]      # superior = SMALLER Y (grid_sample y=-1 → top row = superior)

    # Central/subareolar = within 25 mm of the NIPPLE.
    # The nipple is the most anterior (lowest Z) vertex on the affected breast skin.
    #
    # BUG FIXED: previously used dist_c = dist from breast CENTROID. The centroid of
    # a surface point cloud sits inside the breast volume, ~30-50 mm from any surface
    # vertex, so (dist_c < 25mm) was always False → C = 0 verts → all C draws fell
    # back to UO, silently biasing the distribution to 68% UO instead of 50%.
    breast_sp  = sp[breast]
    nipple_pt  = breast_sp[np.argmin(breast_sp[:, 2])]   # lowest Z = most anterior
    dist_nipple = np.linalg.norm(sp - nipple_pt, axis=1)
    central = breast & (dist_nipple < 25.0)

    code = np.full(len(sp), 'other', dtype=object)
    code[breast &  upper &  outer & ~central] = 'UO'
    code[breast &  upper & ~outer & ~central] = 'UI'
    code[breast & ~upper &  outer & ~central] = 'LO'
    code[breast & ~upper & ~outer & ~central] = 'LI'
    code[central]                             = 'C'
    return code, c, nipple_pt, x_mid, side_pos

qcode, c_aff, nipple_pt, x_mid, side_pos = quadrants(geo)

print(f'Affected breast : {"+" if side_pos else "-"}X of midline')
print(f'Breast centroid  (UO/UI/LO/LI axes): '
      f'x={c_aff[0]:.0f}  y={c_aff[1]:.0f}  z={c_aff[2]:.0f} mm')
print(f'Nipple proxy (min-Z of skin)        : '
      f'x={nipple_pt[0]:.0f}  y={nipple_pt[1]:.0f}  z={nipple_pt[2]:.0f} mm')
print(f'  → outer = X {">" if side_pos else "<"} {c_aff[0]:.0f} mm  (away from sternum)')
print(f'  → upper = Y < {c_aff[1]:.0f} mm  (superior = smaller Y)')
print(f'  → central = within 25 mm of nipple')
print()
for lbl in ['UO','UI','LO','LI','C']:
    print(f'  {lbl:2s} ({QUAD_NAMES[lbl]:20s}): {int((qcode==lbl).sum()):5d} verts  '
          f'[literature: {QUAD_DIST[lbl]*100:.0f}%]')

# ── Literature-based synthetic tumour placement ────────────────────────────────
def plant_synthetic_tumor(geo, qcode, c_aff, rng=None):
    if rng is None:
        rng = np.random.default_rng()
    sp  = geo['surface_pts']
    nrm = geo['vertex_normals']

    labels = list(QUAD_DIST.keys())
    probs  = [QUAD_DIST[l] for l in labels]
    chosen = rng.choice(labels, p=probs)

    mask = (qcode == chosen)
    if mask.sum() < 10:
        chosen = 'UO'; mask = (qcode == 'UO')

    patch_pts = sp[mask];  patch_nrm = nrm[mask]
    c_surf    = patch_pts.mean(0)
    inward    = -patch_nrm.mean(0)
    inward   /= np.linalg.norm(inward) + 1e-9

    d_mm      = float(rng.uniform(10.0, 20.0))
    skin_clr  = float(rng.uniform(5.0,  20.0))
    depth     = d_mm / 2.0 + skin_clr
    pt        = c_surf + depth * inward

    return (dict(x_t_mm=float(pt[0]), y_t_mm=float(pt[1]),
                 z_t_mm=float(pt[2]), d_mm=d_mm),
            chosen, skin_clr)

rng   = np.random.default_rng(42)
plant, chosen_quad, skin_clr = plant_synthetic_tumor(geo, qcode, c_aff, rng)
print(f'\nSynthetic tumour:')
print(f'  Quadrant   : {chosen_quad} — {QUAD_NAMES[chosen_quad]}  (p={QUAD_DIST[chosen_quad]*100:.0f}%)')
print(f'  Diameter   : {plant["d_mm"]:.1f} mm')
print(f'  Skin depth : {skin_clr:.1f} mm')
print(f'  Position   : x={plant["x_t_mm"]:.1f}  y={plant["y_t_mm"]:.1f}  z={plant["z_t_mm"]:.1f} mm')

# Axis check: verify the planted point actually belongs to the claimed quadrant
pt = np.array([plant['x_t_mm'], plant['y_t_mm'], plant['z_t_mm']])
is_outer = (pt[0] > c_aff[0]) if side_pos else (pt[0] < c_aff[0])
is_upper = pt[1] < c_aff[1]
dist_nip = np.linalg.norm(pt - nipple_pt)
expected = {
    'UO': is_upper and is_outer and dist_nip >= 25,
    'UI': is_upper and not is_outer and dist_nip >= 25,
    'LO': not is_upper and is_outer and dist_nip >= 25,
    'LI': not is_upper and not is_outer and dist_nip >= 25,
    'C':  dist_nip < 25,
}
ok = expected.get(chosen_quad, False)
print(f'\n  Axis check: {"upper" if is_upper else "lower"} / '
      f'{"outer" if is_outer else "inner"} / '
      f'dist_nipple={dist_nip:.0f}mm → '
      f'{"✓ matches " + chosen_quad if ok else "⚠ MISMATCH — planted point not in " + chosen_quad}')

# ── Visualisation ─────────────────────────────────────────────────────────────
palette = {'UO':'crimson','UI':'orange','LO':'gold','LI':'khaki',
           'C':'mediumpurple','other':'lightgray'}
vcol = np.array([palette[q] for q in qcode], dtype=object)

fig = go.Figure(go.Mesh3d(
    x=sp[:,0], y=sp[:,1], z=sp[:,2], i=faces[:,0], j=faces[:,1], k=faces[:,2],
    vertexcolor=vcol, flatshading=False, opacity=0.5,
    lighting=dict(ambient=0.6, diffuse=0.8), showscale=False, name=''))

for color, label in [('crimson',      'UO — Upper-Outer   50%'),
                     ('orange',       'UI — Upper-Inner   15%'),
                     ('gold',         'LO — Lower-Outer   11%'),
                     ('khaki',        'LI — Lower-Inner    6%'),
                     ('mediumpurple', 'C  — Central/subareolar 18%'),
                     ('lightgray',    'Contralateral breast')]:
    fig.add_trace(go.Scatter3d(x=[None], y=[None], z=[None], mode='markers',
                               marker=dict(size=8, color=color, symbol='square'),
                               name=label))

# Reference points for visual verification
fig.add_trace(go.Scatter3d(
    x=[nipple_pt[0]], y=[nipple_pt[1]], z=[nipple_pt[2]], mode='markers',
    marker=dict(size=7, color='white', symbol='diamond', line=dict(color='black', width=1)),
    name=f'nipple ({nipple_pt[0]:.0f},{nipple_pt[1]:.0f},{nipple_pt[2]:.0f})'))
fig.add_trace(go.Scatter3d(
    x=[c_aff[0]], y=[c_aff[1]], z=[c_aff[2]], mode='markers',
    marker=dict(size=6, color='cyan', symbol='cross'),
    name=f'centroid ({c_aff[0]:.0f},{c_aff[1]:.0f},{c_aff[2]:.0f})'))

plant_pt = [plant['x_t_mm'], plant['y_t_mm'], plant['z_t_mm']]
fig.add_trace(sphere_mesh(plant_pt, plant['d_mm']/2.0, 'black',
                          name=f'synthetic tumour ({chosen_quad}, d={plant["d_mm"]:.0f}mm)'))
fig.update_layout(
    title=(f'Quadrant map — literature distribution (Senie 1983, Kricker 2007)<br>'
           f'<sup>sampled: {chosen_quad} ({QUAD_NAMES[chosen_quad]}), '
           f'd={plant["d_mm"]:.0f}mm, skin depth={skin_clr:.0f}mm | '
           f'◆=nipple (central pivot), +=centroid (UO/LI pivot) | superior=−Y</sup>'),
    scene=dict(aspectmode='data', camera=ANT_CAM),
    legend=dict(x=0.01, y=0.97, bgcolor='rgba(255,255,255,0.85)',
                bordercolor='gray', borderwidth=1),
    height=660, width=920, margin=dict(l=0, r=0, t=65, b=0))
fig.write_html(str(R.RESULTS_DIR / f'{pid}_quadrant_map.html'))
fig.show()
```

## 6 · Forward model — steady Pennes (FEniCSx)

Paper-faithful weak form with **spatially-varying** perfusion and metabolic heat (tumour vs healthy), reusing the verified dolfinx-0.10 skeleton from `run_cohort.run_fea_forward`. The tumour is a smooth source (Gaussian indicator) so the surface response to $(x,y,z,d)$ is differentiable — important for the LM Jacobian later.

$$ k\nabla^2 T - P(\mathbf{x})\,(T - T_a) + Q(\mathbf{x}) = 0, \qquad P,Q \text{ raised inside the tumour} $$

```python
def pennes_forward(msh_path, beta, geo, const=PAPER):
    # Steady Pennes forward solve. beta: x_t_mm, y_t_mm, z_t_mm, d_mm [, k_h].
    # Returns (T_sol, msh). Spatial perfusion/metabolic fields (tumour vs healthy).
    from mpi4py import MPI
    import dolfinx, dolfinx.mesh
    from dolfinx.io import gmsh as gmshio
    from dolfinx import fem
    from dolfinx.fem.petsc import LinearProblem
    import ufl, numpy as _np

    fea = gmshio.read_from_msh(str(msh_path), MPI.COMM_WORLD, gdim=3)
    msh = fea.mesh
    msh.geometry.x[:] *= 1e-3                       # mm -> m

    Vs = fem.functionspace(msh, ('Lagrange', 1))
    T  = ufl.TrialFunction(Vs); v = ufl.TestFunction(Vs)
    Xc = ufl.SpatialCoordinate(msh)

    k_h    = float(beta.get('k_h', const['k_h']))
    perf_h = const['omega_h'] * const['rho_b'] * const['c_b']
    perf_t = const['omega_t'] * const['rho_b'] * const['c_b']
    Ta, Q_h, h, T_air = const['T_a'], const['Q_h'], const['h'], const['T_inf']

    x_t = beta['x_t_mm']*1e-3; y_t = beta['y_t_mm']*1e-3; z_t = beta['z_t_mm']*1e-3
    r_t = (beta['d_mm']/2.0)*1e-3
    Q_t = float(R.tumor_Qm_from_radius(torch.tensor(beta['d_mm']/2.0)).item())

    d2   = (Xc[0]-x_t)**2 + (Xc[1]-y_t)**2 + (Xc[2]-z_t)**2
    chi  = ufl.exp(-d2 / (r_t**2 + 1e-16))          # smooth tumour indicator
    perf = perf_h + (perf_t - perf_h) * chi
    Q    = Q_h    + (Q_t    - Q_h)    * chi

    # chest-wall Dirichlet: posterior Z-slab (back CHEST_WALL_FRAC of mesh Z range)
    # using facet midpoint Z — avoids the normal-based over-selection on curved surfaces
    fdim = msh.topology.dim - 1
    msh.topology.create_connectivity(fdim, msh.topology.dim)
    ext = dolfinx.mesh.exterior_facet_indices(msh.topology)
    mid = dolfinx.mesh.compute_midpoints(msh, fdim, ext)   # shape (n_ext, 3), in metres
    mz = mid[:, 2]
    mz0, mz1 = float(mz.min()), float(mz.max())
    post   = mz > (mz0 + (1.0 - R.CHEST_WALL_FRAC) * (mz1 - mz0))
    facets = ext[post]
    dofs = fem.locate_dofs_topological(Vs, fdim, facets)
    bc   = fem.dirichletbc(_np.float64(Ta), dofs, Vs)

    a = (k_h*ufl.inner(ufl.grad(T), ufl.grad(v)) + perf*T*v)*ufl.dx + h*T*v*ufl.ds
    L = (perf*Ta + Q)*v*ufl.dx + h*T_air*v*ufl.ds
    try:
        prob = LinearProblem(a, L, bcs=[bc], petsc_options_prefix='lm_',
                             petsc_options={'ksp_type': 'cg', 'pc_type': 'gamg'})
    except TypeError:
        prob = LinearProblem(a, L, bcs=[bc],
                             petsc_options={'ksp_type': 'cg', 'pc_type': 'gamg'})
    return prob.solve(), msh

print('pennes_forward defined.')
```

## 8 · Levenberg–Marquardt inverse — estimating tumour size & location

We **invert** to recover β = [x_t, y_t, z_t, d] by minimising the direct MSE:
$$S(\beta)=\tfrac1N\sum_{i\in\text{skin}}\big(T_i^{\text{meas}}-T_i^{\text{FEM}}(\beta)\big)^2 .$$

**Levenberg–Marquardt** with Marquardt diagonal scaling and **residual-Jacobian** sign convention:
$$\big(J^{\!\top}J+\mu\,\mathrm{diag}(J^{\!\top}J)\big)\,\delta\beta = -J^{\!\top}r, \qquad r=T^{\text{meas}}-T^{\text{FEM}},\quad J_{ij}=\partial r_i/\partial\beta_j .$$
Jacobian by forward finite-differences; μ ×0.3 on improvement, ×3 on rejection.

A `PennesCtx` reads the **3 mm mesh once** and re-solves per β, avoiding repeated mesh I/O per iteration (~12 s overhead per dolfinx read).

**Detection:** if LM drives the tumour **outside the breast** or to a vanishing diameter → benign; a confined, plausible tumour → suspicious/malignant.

```python
# ---- §8.1  Persistent Pennes solver (read 3mm mesh ONCE, re-solve per beta) ----
import ufl
import dolfinx, dolfinx.mesh
from mpi4py import MPI
from dolfinx.io import gmsh as gmshio
from dolfinx import fem
from dolfinx.fem.petsc import LinearProblem
from scipy.spatial import cKDTree

class PennesCtx:
    """Reads the 3 mm mesh + builds the BC and surface-interpolation map ONCE;
    .solve(beta) only reassembles the UFL forms — no mesh I/O overhead per call."""
    def __init__(self, msh_path, geo, const=PAPER):
        fea = gmshio.read_from_msh(str(msh_path), MPI.COMM_WORLD, gdim=3)
        self.msh = fea.mesh
        self.msh.geometry.x[:] *= 1e-3
        self.V  = fem.functionspace(self.msh, ('Lagrange', 1))
        self.Xc = ufl.SpatialCoordinate(self.msh)
        self.const = const
        # chest-wall Dirichlet: posterior Z-slab on facet midpoints
        fdim = self.msh.topology.dim - 1
        self.msh.topology.create_connectivity(fdim, self.msh.topology.dim)
        ext = dolfinx.mesh.exterior_facet_indices(self.msh.topology)
        mid = dolfinx.mesh.compute_midpoints(self.msh, fdim, ext)
        mz = mid[:, 2]; mz0, mz1 = float(mz.min()), float(mz.max())
        post   = mz > (mz0 + (1.0 - R.CHEST_WALL_FRAC) * (mz1 - mz0))
        facets = ext[post]
        self.bc = fem.dirichletbc(np.float64(const['T_a']),
                                  fem.locate_dofs_topological(self.V, fdim, facets), self.V)
        _, self.sidx = cKDTree(self.msh.geometry.x * 1e3).query(geo['surface_pts'], k=3)
        self.n_nodes = self.msh.geometry.x.shape[0]

    def solve(self, beta):
        c = self.const
        T = ufl.TrialFunction(self.V); v = ufl.TestFunction(self.V)
        k_h    = float(beta.get('k_h', c['k_h']))
        perf_h = c['omega_h'] * c['rho_b'] * c['c_b']
        perf_t = c['omega_t'] * c['rho_b'] * c['c_b']
        Ta, Q_h, h, T_air = c['T_a'], c['Q_h'], c['h'], c['T_inf']
        x_t = beta['x_t_mm']*1e-3; y_t = beta['y_t_mm']*1e-3; z_t = beta['z_t_mm']*1e-3
        dmm = max(float(beta['d_mm']), 1.0); r_t = (dmm / 2.0) * 1e-3
        Q_t = float(R.tumor_Qm_from_radius(torch.tensor(dmm / 2.0)).item())
        d2  = (self.Xc[0]-x_t)**2 + (self.Xc[1]-y_t)**2 + (self.Xc[2]-z_t)**2
        chi = ufl.exp(-d2 / (r_t**2 + 1e-16))
        perf = perf_h + (perf_t - perf_h) * chi
        Q    = Q_h    + (Q_t    - Q_h)    * chi
        a = (k_h*ufl.inner(ufl.grad(T), ufl.grad(v)) + perf*T*v)*ufl.dx + h*T*v*ufl.ds
        L = (perf*Ta + Q)*v*ufl.dx + h*T_air*v*ufl.ds
        try:
            prob = LinearProblem(a, L, bcs=[self.bc], petsc_options_prefix='inv_',
                                 petsc_options={'ksp_type': 'cg', 'pc_type': 'gamg'})
        except TypeError:
            prob = LinearProblem(a, L, bcs=[self.bc],
                                 petsc_options={'ksp_type': 'cg', 'pc_type': 'gamg'})
        sol = prob.solve()
        return sol.x.array[self.sidx].mean(axis=1)

apply_constants()
ctx = PennesCtx(msh_path, geo)   # 3 mm mesh — same as forward solve
print(f'PennesCtx ready: {ctx.n_nodes:,} nodes  (3 mm mesh)')
```

```python
# ---- §8.2  Levenberg-Marquardt loop ----------------------------------------
#
# TWO MODES (flip SYNTHETIC_TEST):
#   True  → target = ctx.solve(plant)[skin]     (same 3mm mesh → zero floor)
#   False → target = geo['T_measured'][skin]     (real NeRF IR)
#
# INITIALIZATION — differential hotspot:
#   The ABSOLUTE temperature hotspot is dominated by chest-wall proximity
#   (Dirichlet 37°C pulls the inferior skin to ~35°C regardless of the tumour).
#   The tumour signal is only visible in ΔT = T_target − T_healthy.
#   The hotspot of ΔT is at the skin projection of the tumour → good XY start.
#   Cost: 1 extra FEM solve for the healthy baseline.
#
SYNTHETIC_TEST = True   # ← flip to False for real-patient run

# Z-slab skin mask
_z_s  = geo['surface_pts'][:, 2]
_z0_s, _z1_s = float(_z_s.min()), float(_z_s.max())
skin  = _z_s < (_z0_s + (1.0 - R.CHEST_WALL_FRAC) * (_z1_s - _z0_s))

if SYNTHETIC_TEST:
    T_lm_target = ctx.solve(plant)[skin].astype(float)
    print('MODE: SYNTHETIC VALIDATION  (target = ctx.solve(plant), same 3mm mesh)')
    print(f'  Known answer : {({k: round(plant[k],1) for k in ["x_t_mm","y_t_mm","z_t_mm","d_mm"]})}')
    print(f'  Quadrant     : {chosen_quad} — {QUAD_NAMES[chosen_quad]}')
else:
    T_lm_target = geo['T_measured'][skin].astype(float)
    print('MODE: REAL PATIENT INVERSE')
    print(f'  T_measured   : {T_lm_target.min():.2f}–{T_lm_target.max():.2f} C')

bbmin = np.asarray(geo['bbox_min'], float)
bbmax = np.asarray(geo['bbox_max'], float)

PNAMES = ['x_t_mm', 'y_t_mm', 'z_t_mm', 'd_mm']
STEP   = {'x_t_mm': 3.0, 'y_t_mm': 3.0, 'z_t_mm': 3.0, 'd_mm': 2.0}

def vec2beta(x):
    b = {k: float(v) for k, v in zip(PNAMES, x)}
    b['d_mm'] = max(b['d_mm'], 1.0)
    b['k_h']  = PAPER['k_h']
    return b

def resid(x):
    # r = target - T_model(x);  J[:,j] = ∂r/∂x_j  (residual Jacobian)
    return T_lm_target - ctx.solve(vec2beta(x))[skin]

def cost(r):
    return float(np.mean(r ** 2))

# ── Differential hotspot initialisation (1 extra FEM solve) ──────────────────
# T_healthy: same mesh, tumour pushed 1000 mm outside bbox → chi ≈ 0 everywhere
_hlth_beta = dict(x_t_mm=float(bbmax[0]) + 1000.0,
                  y_t_mm=0.0, z_t_mm=0.0, d_mm=1.0, k_h=PAPER['k_h'])
T_hlth  = ctx.solve(_hlth_beta)[skin]
n_solves = 1

dT       = T_lm_target - T_hlth          # tumour's thermal contribution only
hot_abs  = geo['surface_pts'][skin][int(np.argmax(T_lm_target))]
hot_diff = geo['surface_pts'][skin][int(np.argmax(dT))]
z_init   = float(0.5 * (bbmin[2] + bbmax[2]))

x = np.array([hot_diff[0], hot_diff[1], z_init, 12.0])

print(f'\nAbsolute hotspot  : ({hot_abs[0]:.0f}, {hot_abs[1]:.0f}) mm  '
      f'← dominated by chest-wall BC')
print(f'Differential hotspot (ΔT): ({hot_diff[0]:.0f}, {hot_diff[1]:.0f}) mm  '
      f'← tumour signal only')
if SYNTHETIC_TEST:
    print(f'True tumour XY    : ({plant["x_t_mm"]:.0f}, {plant["y_t_mm"]:.0f}) mm')
print(f'Initial guess : {dict(zip(PNAMES, np.round(x, 1)))}')

# ── LM iterations ─────────────────────────────────────────────────────────────
r = resid(x); S = cost(r); n_solves += 1
mu = 1e-2; MAX_ITER = 25
print(f'start  RMS={np.sqrt(S):.4f} C')

# Initialise history dict (used by §9.4 convergence plot — zero extra FEM cost)
lm_hist = {k: [float(x[i])] for i, k in enumerate(PNAMES)}
lm_hist['RMS']  = [float(np.sqrt(S))]
lm_hist['mu']   = [float(mu)]
lm_hist['iter'] = [0]

for it in range(1, MAX_ITER + 1):
    # Forward-difference Jacobian: J[:,j] = (r(x+h_j) − r(x)) / h_j = ∂r/∂x_j
    J = np.empty((r.size, len(PNAMES)))
    for j, name in enumerate(PNAMES):
        xp = x.copy(); xp[j] += STEP[name]
        J[:, j] = (resid(xp) - r) / STEP[name]
        n_solves += 1

    JtJ = J.T @ J
    Jtr = J.T @ r

    accepted = False
    for _ in range(8):
        A  = JtJ + mu * np.diag(np.diag(JtJ)) + 1e-9 * np.eye(len(x))
        # Residual-Jacobian LM normal eq: (J^T J + mu D) δx = −J^T r
        dx = np.linalg.solve(A, -Jtr)
        r_n = resid(x + dx); S_n = cost(r_n); n_solves += 1
        if S_n < S:
            x, r, S = x + dx, r_n, S_n
            mu = max(mu * 0.3, 1e-9); accepted = True; break
        mu = min(mu * 3.0, 1e8)

    # Log to history
    for _hi, _hk in enumerate(PNAMES):
        lm_hist[_hk].append(float(x[_hi]))
    lm_hist['RMS'].append(float(np.sqrt(S)))
    lm_hist['mu'].append(float(mu))
    lm_hist['iter'].append(it)

    tag = '' if accepted else '  [rejected]'
    print(f'iter {it:2d}  RMS={np.sqrt(S):.4f} C  mu={mu:.1e}  '
          f'|dx|={np.linalg.norm(dx):5.2f}  '
          f'{dict(zip(PNAMES, np.round(x, 1)))}{tag}')
    if not accepted:
        print('  no improvement → stop'); break
    if np.linalg.norm(dx) < 0.05 or S < 1e-6:
        print('  converged'); break

beta_hat = vec2beta(x)
print(f'\nLM finished : {n_solves} FEM solves   final RMS={np.sqrt(S):.4f} C')
print(f'beta_hat    : {({k: round(beta_hat[k],1) for k in PNAMES})}')
if SYNTHETIC_TEST:
    err     = {k: round(beta_hat[k] - plant[k], 1) for k in PNAMES}
    pos_err = np.sqrt(err['x_t_mm']**2 + err['y_t_mm']**2 + err['z_t_mm']**2)
    print(f'plant       : {({k: round(plant[k],1) for k in PNAMES})}')
    print(f'error       : {err}')
    print(f'position error: {pos_err:.1f} mm')
```

```python
# ---- §8.3  Detection ("boundary condition") + result figure ----
centre  = np.array([beta_hat['x_t_mm'], beta_hat['y_t_mm'], beta_hat['z_t_mm']])
d_hat   = beta_hat['d_mm']
pad     = 5.0
outside = bool(np.any(centre < bbmin - pad) or np.any(centre > bbmax + pad))
tiny    = d_hat < 4.0
if outside or tiny:
    verdict = 'BENIGN  (no confined source: tumour driven outside / to ~0 diameter)'
else:
    verdict = 'SUSPICIOUS → MALIGNANT  (confined, plausible tumour recovered)'

print('Detection')
print(f"  recovered centre : ({centre[0]:.0f}, {centre[1]:.0f}, {centre[2]:.0f}) mm")
print(f"  breast bbox      : X[{bbmin[0]:.0f},{bbmax[0]:.0f}] "
      f"Y[{bbmin[1]:.0f},{bbmax[1]:.0f}] Z[{bbmin[2]:.0f},{bbmax[2]:.0f}]")
print(f"  recovered d      : {d_hat:.1f} mm    inside={not outside}")
print(f"  VERDICT          : {verdict}")
if SYNTHETIC_TEST:
    print(f"  (synthetic — known plant at "
          f"x={plant['x_t_mm']:.0f} y={plant['y_t_mm']:.0f} "
          f"z={plant['z_t_mm']:.0f} d={plant['d_mm']:.1f})")

T_fit = ctx.solve(beta_hat)
fig = go.Figure(go.Mesh3d(
    x=sp[:,0], y=sp[:,1], z=sp[:,2], i=faces[:,0], j=faces[:,1], k=faces[:,2],
    intensity=T_fit, colorscale='Rainbow', intensitymode='vertex',
    colorbar=dict(title='C'), flatshading=False, opacity=0.35,
    lighting=dict(ambient=0.6, diffuse=0.8)))
if not outside:
    fig.add_trace(sphere_mesh(list(centre), max(d_hat, 2.0)/2.0, 'lime',
                              name='recovered tumour'))
if SYNTHETIC_TEST:
    fig.add_trace(sphere_mesh([plant['x_t_mm'], plant['y_t_mm'], plant['z_t_mm']],
                              plant['d_mm']/2.0, 'cyan', opacity=0.5,
                              name='planted (ground truth)'))
fig.update_layout(
    title=(f'LM inverse result<br>'
           f'<sup>final RMS={np.sqrt(S):.3f} C | d_hat={d_hat:.1f} mm | '
           f'{verdict.split("  ")[0]}</sup>'),
    scene=dict(aspectmode='data', camera=ANT_CAM),
    height=620, width=860, margin=dict(l=0, r=0, t=60, b=0))
fig.write_html(str(R.RESULTS_DIR / f'{pid}_LM_inverse_result.html'))
fig.show()
```

```python
# ---- §8.4  Identifiability - Jacobian sensitivity & rank at the solution ----
Jf = np.empty((r.size, len(PNAMES)))
for j, name in enumerate(PNAMES):
    xp = x.copy(); xp[j] += STEP[name]
    Jf[:, j] = (resid(xp) - r) / STEP[name]
# scaled sensitivity: RMS surface-temp change for a one-STEP move in each parameter
sens = np.linalg.norm(Jf, axis=0) * np.array([STEP[k] for k in PNAMES]) / np.sqrt(r.size)
U, sv, Vt = np.linalg.svd(Jf, full_matrices=False)
rank = int(np.sum(sv > sv[0]*1e-3)) if sv[0] > 0 else 0

print('Identifiability (at the LM solution)')
for name, s in zip(PNAMES, sens):
    flag = 'identifiable' if s > 1e-3 else 'NOT identifiable (flat)'
    print(f'  dRMS / d {name:8s} ~ {s:7.4f} C/step   {flag}')
print(f'  singular values : {np.array2string(sv, precision=3)}')
print(f'  numerical rank  : {rank} / {len(PNAMES)}   '
      f'(paper: 5/8 with 8 views; we have 5 views)')
print('\nParameters with ~0 sensitivity (typically depth z_t) are not resolvable')
print('from this surface data alone -> their recovered values are prior-dominated,')
print('exactly the limitation Gonzalez-Hernandez / Kandlikar et al. report.')
```

```python
# ── §9.1  Mesh h-convergence ─────────────────────────────────────────────────
# Generates and solves at 5, 4, 3, 2 mm (and optionally 1 mm).
# Reference = finest mesh available.  P1 Lagrange → O(h²) convergence expected.
#
# Cost: 5mm ~2s, 4mm ~5s, 3mm reused from §7, 2mm ~42s, 1mm ~310s (disabled).
# Requires: stl_path, plant, geo, skin (§8.2), T_surf (§7), ctx (§8.1).

SKIP_1MM = True    # ← set False to add 1mm (~310 s, ~3.5 M nodes)

_mesh_sizes = [5.0, 4.0, 3.0, 2.0] + ([] if SKIP_1MM else [1.0])
_T_surfs    = {}   # h → skin temperatures (skin mask from §8.2)
_n_nd       = {}   # h → node count
_t_solve    = {}   # h → wall-clock solve time (s)
_msh_paths  = {}   # h → Path to .msh file

print(f'§9.1  Mesh h-convergence  (h = {_mesh_sizes} mm)')
for _h in _mesh_sizes:
    _msh_h = R.STL_DIR / f'{pid}_{_h:g}mm.msh'
    _msh_paths[_h] = _msh_h

    # Reuse §7 forward solve at 3mm (same mesh + same interpolation → no comparison bias)
    if _h == 3.0:
        print(f'  {_h:g}mm: reusing §7 forward result  {ctx.n_nodes:,} nodes')
        _T_surfs[3.0]  = T_surf[skin].astype(float)
        _n_nd[3.0]     = ctx.n_nodes
        _t_solve[3.0]  = 0.0
        continue

    if not _msh_h.exists():
        print(f'  generating {_h:g}mm mesh...', end=' ', flush=True)
        R.stl_to_tet_mesh(str(stl_path), str(_msh_h), mesh_size_mm=_h)
        print('done')

    print(f'  solving {_h:g}mm...', end=' ', flush=True)
    _t0 = time.time()
    _Tsol, _mobj = pennes_forward(_msh_h, plant, geo)
    _, _, _, _Th = R.compute_fea_residual(_Tsol, _mobj, geo['surface_pts'], geo['T_measured'])
    _t_solve[_h]  = time.time() - _t0
    _n_nd[_h]     = _mobj.geometry.x.shape[0]
    _T_surfs[_h]  = _Th[skin].astype(float)
    print(f'{_n_nd[_h]:,} nodes  {_t_solve[_h]:.1f}s')

# Convenience aliases used by §9.3
msh_5mm = _msh_paths[5.0]
if 2.0 in _msh_paths: msh_2mm = _msh_paths[2.0]

# Reference = finest mesh available
_h_ref  = min(_mesh_sizes)
_T_ref  = _T_surfs[_h_ref]
_signal = float(_T_surfs[3.0].max() - _T_surfs[3.0].mean())

print(f'\n  Reference h = {_h_ref}mm ({_n_nd[_h_ref]:,} nodes)')
print(f'  Tumour signal (3mm, peak−mean) = {_signal:.4f} °C')
print(f'\n  {"h (mm)":<8} {"nodes":>10} {"t_solve (s)":>12} {"max|ΔT| (°C)":>14} {"RMS|ΔT| (°C)":>14}')
print('  ' + '─'*62)

_max_err = {}; _rms_err = {}
for _h in sorted(_mesh_sizes, reverse=True):
    _dT = _T_surfs[_h] - _T_ref
    _max_err[_h] = float(np.abs(_dT).max())
    _rms_err[_h] = float(np.sqrt(np.mean(_dT**2)))
    _ref_tag = ' ← ref' if _h == _h_ref else ''
    print(f'  {_h:<8.1f} {_n_nd[_h]:>10,} {_t_solve[_h]:>12.1f} '
          f'{_max_err[_h]:>14.4f} {_rms_err[_h]:>14.4f}{_ref_tag}')

# Convergence rate — log-log slope (exclude reference)
_h_fit = sorted([_h for _h in _mesh_sizes if _h != _h_ref and _max_err[_h] > 1e-8])
if len(_h_fit) >= 2:
    _slope = np.polyfit(np.log(_h_fit),
                        np.log([_max_err[_h] for _h in _h_fit]), 1)[0]
    print(f'\n  Measured convergence rate  p ≈ {_slope:.2f}')
    print(f'  Theory (P1 Lagrange, smooth solution): p = 2')

print(f'\n  3mm error/signal = {_max_err[3.0]/_signal:.2%} → h=3mm sufficient')

# ── Plot 1: log-log convergence curve (620×860) ────────────────────────────
_h_c  = sorted([_h for _h in _mesh_sizes if _h != _h_ref], reverse=True)
_errs = [_max_err[_h] for _h in _h_c]

fig91a = go.Figure()
fig91a.add_trace(go.Scatter(
    x=_h_c, y=_errs, mode='lines+markers',
    marker=dict(size=10, color='royalblue'), line=dict(width=2, color='royalblue'),
    name='max|ΔT| (measured)'))

# O(h²) reference line anchored at coarsest point
_h2_x = np.array([min(_h_c), max(_h_c)])
_C2   = _max_err[max(_h_c)] / max(_h_c) ** 2
fig91a.add_trace(go.Scatter(
    x=_h2_x.tolist(), y=(_C2 * _h2_x ** 2).tolist(), mode='lines',
    line=dict(dash='dash', color='gray', width=1.5), name='O(h²) reference'))

# 1% of tumour signal floor
_floor = _signal * 0.01
fig91a.add_hline(y=_floor, line_dash='dot', line_color='seagreen',
    annotation_text=f'1% of signal ({_floor:.4f}°C)',
    annotation_position='bottom right')

fig91a.update_layout(
    title=(f'§9.1a Mesh convergence — max|ΔT| vs h  (ref = {_h_ref}mm)<br>'
           f'<sup>P1 Lagrange → O(h²). Dot = 1% tumour signal floor. '           f'{"1mm omitted (SKIP_1MM=True)" if SKIP_1MM else "1mm included"}</sup>'),
    xaxis=dict(title='Element size h (mm)', type='log'),
    yaxis=dict(title='max|ΔT| (°C)',        type='log'),
    height=620, width=860)
fig91a.show()

# ── Plot 2: T(h) vs T(ref) scatter (620×860) ──────────────────────────────
_clr91 = {5.0: 'tomato', 4.0: 'orange', 3.0: 'royalblue', 2.0: 'seagreen', 1.0: 'purple'}
fig91b = go.Figure()
for _h in _h_c:
    fig91b.add_trace(go.Scatter(
        x=_T_ref, y=_T_surfs[_h], mode='markers',
        marker=dict(size=3, color=_clr91.get(_h, 'gray'), opacity=0.35),
        name=f'h={_h:g}mm  (max|ΔT|={_max_err[_h]:.4f}°C)'))
_Tr = [float(_T_ref.min()) - 0.05, float(_T_ref.max()) + 0.05]
fig91b.add_trace(go.Scatter(x=_Tr, y=_Tr, mode='lines',
    line=dict(color='black', dash='dash', width=1.5), name='y = x (ideal)'))
fig91b.update_layout(
    title=(f'§9.1b T(h) vs T({_h_ref}mm ref) at skin vertices'),
    xaxis_title=f'T at {_h_ref}mm (°C)', yaxis_title='T at h (°C)',
    height=620, width=860)
fig91b.show()
```

```python
# ── §9.2  FD step-size sensitivity ──────────────────────────────────────────
# At the converged LM solution, sweep h_d and measure |∂T/∂d|_RMS.
# Optimal h_d lies in the flat plateau where truncation ≈ discretisation noise.
# 6 extra FEM solves (~78 s).
# Requires: ctx, x (converged params from §8.2), skin, vec2beta, STEP.

print('§9.2 FD step-size sensitivity  (d_mm at converged β)')
h_d_vals = [0.5, 1.0, 2.0, 3.0, 5.0, 8.0]
T0_lm = ctx.solve(vec2beta(x))[skin]
J_rms = []
for h_d in h_d_vals:
    xp = x.copy(); xp[3] += h_d
    J_col = (ctx.solve(vec2beta(xp))[skin] - T0_lm) / h_d
    J_rms.append(float(np.sqrt(np.mean(J_col**2))))
    print(f'  h_d={h_d:5.1f}mm   |∂T/∂d| RMS = {J_rms[-1]:.5f} °C/mm')

chosen_idx = h_d_vals.index(STEP['d_mm'])
plateau_mean = float(np.mean([J_rms[i] for i, h in enumerate(h_d_vals) if 1.0 <= h <= 5.0]))

fig92 = go.Figure()
fig92.add_trace(go.Scatter(
    x=h_d_vals, y=J_rms, mode='lines+markers',
    marker=dict(size=9, color='tomato'), line=dict(width=2, color='tomato'),
    name='|∂T/∂d|'))
fig92.add_vline(x=STEP['d_mm'], line_dash='dash', line_color='green', line_width=2,
    annotation_text=f'chosen h_d={STEP["d_mm"]}mm',
    annotation_position='top right')
fig92.update_layout(
    title=('§9.2 FD step-size sensitivity — ∂T/∂d at converged LM solution<br>'
           '<sup>Flat plateau = optimal range. Chosen h_d=2mm (green) is in plateau.</sup>'),
    xaxis_title='FD step h_d (mm)', yaxis_title='|∂T/∂d| RMS (°C/mm)',
    height=620, width=860)
fig92.show()
print(f'\n  chosen h_d={STEP["d_mm"]}mm → {J_rms[chosen_idx]:.5f} °C/mm')
print(f'  plateau mean (h=1–5mm):    {plateau_mean:.5f} °C/mm')
```

```python
# ── §9.3  CG iteration count vs mesh size ────────────────────────────────────
# Solve at 5mm and 3mm; capture KSP iteration count from PETSc after solve.
# Confirms GAMG gives near mesh-independent convergence.
# 2 extra FEM solves (~16 s). Requires msh_5mm from §9.1.

from mpi4py import MPI as _MPI
import dolfinx as _dfx, dolfinx.mesh as _dfxm
from dolfinx.io import gmsh as _gmshio
from dolfinx import fem as _fem
from dolfinx.fem.petsc import LinearProblem as _LP
import ufl as _ufl

def _pennes_count(msh_path_, beta_):
    """Forward Pennes solve; returns (n_cg_iters, n_nodes)."""
    fea_ = _gmshio.read_from_msh(str(msh_path_), _MPI.COMM_WORLD, gdim=3)
    m = fea_.mesh; m.geometry.x[:] *= 1e-3
    Vs = _fem.functionspace(m, ('Lagrange', 1))
    T_ = _ufl.TrialFunction(Vs); v_ = _ufl.TestFunction(Vs)
    Xc = _ufl.SpatialCoordinate(m)
    ph = PAPER['omega_h']*PAPER['rho_b']*PAPER['c_b']
    pt = PAPER['omega_t']*PAPER['rho_b']*PAPER['c_b']
    Ta, Qh, h_, Ta_ = PAPER['T_a'], PAPER['Q_h'], PAPER['h'], PAPER['T_inf']
    xt = beta_['x_t_mm']*1e-3; yt = beta_['y_t_mm']*1e-3; zt = beta_['z_t_mm']*1e-3
    rt = (beta_['d_mm']/2)*1e-3
    Qt = float(R.tumor_Qm_from_radius(torch.tensor(beta_['d_mm']/2)).item())
    d2 = (Xc[0]-xt)**2+(Xc[1]-yt)**2+(Xc[2]-zt)**2
    chi = _ufl.exp(-d2/(rt**2+1e-16))
    P = ph+(pt-ph)*chi; Q = Qh+(Qt-Qh)*chi
    fdim = m.topology.dim-1; m.topology.create_connectivity(fdim, m.topology.dim)
    ext = _dfxm.exterior_facet_indices(m.topology)
    mz = _dfxm.compute_midpoints(m, fdim, ext)[:,2]
    mz0,mz1 = mz.min(),mz.max()
    post = mz > (mz0+(1-R.CHEST_WALL_FRAC)*(mz1-mz0))
    bc = _fem.dirichletbc(np.float64(Ta),
                          _fem.locate_dofs_topological(Vs, fdim, ext[post]), Vs)
    a = (PAPER['k_h']*_ufl.inner(_ufl.grad(T_),_ufl.grad(v_))+P*T_*v_)*_ufl.dx+h_*T_*v_*_ufl.ds
    L = (P*Ta+Q)*v_*_ufl.dx+h_*Ta_*v_*_ufl.ds
    try:
        prob = _LP(a, L, bcs=[bc], petsc_options_prefix='cnt_',
                   petsc_options={'ksp_type':'cg','pc_type':'gamg'})
    except TypeError:
        prob = _LP(a, L, bcs=[bc], petsc_options={'ksp_type':'cg','pc_type':'gamg'})
    prob.solve()
    try:
        n_it = int(prob.solver.getIterationNumber())
    except Exception:
        n_it = -1   # fallback if dolfinx API differs
    return n_it, int(m.geometry.x.shape[0])

print('§9.3 CG iteration count vs mesh size (GAMG)')
cg_res = {}
for lbl_, path_ in [('5mm', msh_5mm), ('3mm', msh_path)]:
    print(f'  solving at {lbl_}...', end=' ', flush=True)
    n_it_, n_nd_ = _pennes_count(path_, plant)
    cg_res[lbl_] = (n_it_, n_nd_)
    print(f'{n_it_} CG iters  |  {n_nd_:,} nodes')

lbls_ = list(cg_res)
itrs_ = [cg_res[k][0] for k in lbls_]
nds_  = [cg_res[k][1] for k in lbls_]

fig93 = go.Figure()
fig93.add_trace(go.Bar(x=lbls_, y=itrs_, text=itrs_, textposition='outside',
    marker_color=['steelblue','royalblue'], width=0.4))
for i, (lbl_, nd_) in enumerate(zip(lbls_, nds_)):
    fig93.add_annotation(x=lbl_, y=itrs_[i]-1,
        text=f'{nd_:,} nodes', showarrow=False, font=dict(size=10, color='white'))
fig93.update_layout(
    title=('§9.3 CG iteration count vs mesh size (GAMG preconditioner)<br>'
           '<sup>GAMG: O(log 1/h) iterations — far better than unpreconditioned O(1/h)</sup>'),
    xaxis_title='Mesh element size', yaxis_title='CG iterations',
    yaxis=dict(range=[0, max(itrs_)*1.4]),
    height=620, width=860)
fig93.show()
if itrs_[0] > 0:
    nd_ratio = float(np.sqrt(nds_[1]/nds_[0]))
    it_ratio = itrs_[1]/itrs_[0]
    print(f'\n  3mm/5mm node ratio (sqrt): {nd_ratio:.2f}x  — unpreconditioned CG would scale this way')
    print(f'  3mm/5mm iter ratio (GAMG): {it_ratio:.2f}x  — GAMG compresses the growth')
```
